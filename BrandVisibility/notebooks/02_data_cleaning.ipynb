```python
import pandas as pd
import numpy as np
from pathlib import Path

# Locate the project root
current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_root = current_dir.parent
elif (current_dir / "data" / "processed").exists():
    project_root = current_dir
else:
    project_root = current_dir.parent

processed_dir = project_root / "data" / "processed"

# Load the updated combined dataset
combined_path = processed_dir / "brand_visibility_combined.csv"
df_features = pd.read_csv(combined_path)

# Extract brand from the beginning of the title
known_brands = [
    "Apple", "Samsung", "HP", "Dell", "Lenovo",
    "Sony", "LG", "Boat", "Nike", "Adidas", "Puma",
    "Realme", "OnePlus", "Xiaomi", "Redmi", "Asus",
    "Acer", "JBL", "Canon", "Whirlpool", "Bosch",
    "Philips", "Havells", "Prestige"
]

def extract_brand(title):
    if pd.isna(title):
        return "Unknown"

    title = str(title).strip()

    for brand in sorted(known_brands, key=len, reverse=True):
        if title.lower().startswith(brand.lower()):
            return brand

    return "Unknown"

df_features["brand"] = df_features["title"].apply(extract_brand)

# Create price categories
def categorize_price(price):
    if pd.isna(price) or price <= 0:
        return "Unknown"
    if price < 1000:
        return "Budget"
    if price < 5000:
        return "Mid-Range"
    if price < 20000:
        return "Premium"
    return "High-End"

df_features["price_range"] = df_features["price"].apply(categorize_price)

# Recalculate visibility score from search position
position = pd.to_numeric(df_features["position"], errors="coerce")

df_features["visibility_score"] = np.where(
    position.notna() & (position >= 1),
    np.clip(100 - ((position - 1) / 39 * 100), 0, 100),
    np.nan
)

# Calculate discount only when valid old and current prices exist
current_price = pd.to_numeric(df_features["price"], errors="coerce")
old_price = pd.to_numeric(df_features["old_price"], errors="coerce")

valid_discount = (
    current_price.gt(0)
    & old_price.gt(current_price)
    & old_price.notna()
)

df_features["discount_percent"] = np.nan

df_features.loc[valid_discount, "discount_percent"] = (
    (old_price[valid_discount] - current_price[valid_discount])
    / old_price[valid_discount]
    * 100
)

# Save the updated dataset
features_path = processed_dir / "brand_visibility_features.csv"

df_features.to_csv(
    features_path,
    index=False,
    encoding="utf-8-sig"
)

print("Updated feature dataset saved!")
print("Rows and columns:", df_features.shape)
print("Old prices available:", df_features["old_price"].notna().sum())
print("Valid discounts:", df_features["discount_percent"].notna().sum())
print("Saved to:", features_path)
```
