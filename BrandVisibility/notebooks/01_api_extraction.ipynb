```python
import os
import json
import requests
from pathlib import Path
from dotenv import load_dotenv

# Load your API key
load_dotenv(env_path, override=True)
api_key = os.getenv("SERPAPI_KEY")

# Set up the API request
params = {
    "engine": "google_shopping",
    "q": "iPhone 16",
    "api_key": api_key,
    "gl": "us",
    "hl": "en"
}

# Send the request
response = requests.get(
    "https://serpapi.com/search.json",
    params=params,
    timeout=60
)

print("HTTP status:", response.status_code)
response.raise_for_status()

data = response.json()

# Check for API errors
if "error" in data:
    raise RuntimeError(data["error"])

# Save the response
project_root = env_path.parent
raw_dir = project_root / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)

output_file = raw_dir / "api_raw_data.json"

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)

products = data.get("shopping_results", [])

print("Success! Data saved.")
print("File location:", output_file)
print("Products returned:", len(products))
```

```python
import pandas as pd

df_api = pd.json_normalize(products)

print("Dataset shape:", df_api.shape)
display(df_api.head())
```

```python
import pandas as pd
from pathlib import Path

# Convert the API product results into a DataFrame
df_api = pd.json_normalize(products)

# Display the dataset information
print("Number of rows:", df_api.shape[0])
print("Number of columns:", df_api.shape[1])

# Save the API product data as CSV
csv_file = raw_dir / "api_raw_data.csv"
df_api.to_csv(csv_file, index=False, encoding="utf-8-sig")

print("\nCSV saved successfully!")
print("File location:", csv_file)

display(df_api.head())
```
